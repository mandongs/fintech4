```python
# 라이브러리 import
import ast
import os
import re
from datetime import date, datetime
from pathlib import Path

import faiss
import gradio as gr
import numpy as np
import pandas as pd
import tiktoken
from dotenv import load_dotenv
from google import genai
from google.genai import types
```

```python
# 기본 설정
BASE_DIR = Path.cwd()

POLICY_PATH = BASE_DIR / "youth_policy_final.csv"
EMBED_PATH = BASE_DIR / "policy_embeddings.csv"
FAISS_PATH = BASE_DIR / "policy_vector_index.faiss"

load_dotenv(BASE_DIR / ".env")

API_KEY = os.getenv("google_api_key")

EMBEDDING_MODEL = "gemini-embedding-001"
GENERATION_MODEL = "gemini-3-flash-preview"
EMBED_ENCODING = "cl100k_base"
MAX_CONTEXT_TOKENS = 12000

tokenizer = tiktoken.get_encoding(EMBED_ENCODING)

print("BASE_DIR:", BASE_DIR)
print("POLICY_PATH exists:", POLICY_PATH.exists())
print("API KEY loaded:", bool(API_KEY))
```

```python
def safe_str(x):
    if pd.isna(x):
        return ""
    return str(x).strip()


def numeric(val, default=0.0):
    try:
        if pd.isna(val):
            return default
        return float(val)
    except Exception:
        return default


def compute_age(birth_date: str) -> int:
    try:
        birth = pd.to_datetime(birth_date).date()
    except Exception:
        return 0
    today = date.today()
    return today.year - birth.year - ((today.month, today.day) < (birth.month, birth.day))


def province_from_region(region: str) -> str:
    region = safe_str(region)
    if region.startswith("서울"):
        return "서울"
    if region.startswith("경기"):
        return "경기"
    if region.startswith("인천"):
        return "인천"
    return region[:2]


def get_client():
    if not API_KEY:
        raise RuntimeError(
            "GOOGLE_API_KEY 또는 google_api_key 환경변수가 필요합니다. "
            ".env 파일에 API 키를 넣어주세요."
        )
    return genai.Client(api_key=API_KEY)
```

```python
# 청년정책 데이터 로드
def load_policy_data():
    if not POLICY_PATH.exists():
        raise FileNotFoundError(f"정책 데이터 파일이 없습니다: {POLICY_PATH}")
    policies = pd.read_csv(POLICY_PATH, encoding="utf-8-sig")
    return policies


policies_df = load_policy_data()
print(policies_df.shape)
policies_df.head()
```

```python
# 사용자 입력값을 시스템용 user 객체로 변환
def build_user_from_form(
    gender,
    birth_date,
    education,
    region,
    monthly_income,
    annual_income,
    house_owner,
    social_support,
    spending_type,
    marital_status,
    employment_status,
    years_of_service,
    household_type,
    is_sme_employee,
    asset_total,
    financial_asset,
    debt_status,
    housing_type,
    income_type,
):
    age = compute_age(str(birth_date))
    province = province_from_region(region)

    school_code_map = {
        "고교 졸업": "0049004",
        "대학 재학": "0049005",
        "대졸 예정": "0049006",
        "대학 졸업": "0049007",
        "석박사": "0049008",
        "기타": "0049009",
    }
    school_code = school_code_map.get(education, "0049010")

    job_code_map = {
        "취업": "0013001",
        "자영업": "0013002",
        "취업준비중": "0013003",
        "학생": "0013003",
        "프리랜서": "0013004",
        "일용근로": "0013005",
        "예비창업": "0013006",
        "단기근로": "0013007",
        "영농종사": "0013008",
        "기타": "0013009",
    }
    job_code = job_code_map.get(employment_status, "0013009")

    if marital_status == "기혼":
        mrg_code = "0055001"
    elif marital_status == "미혼":
        mrg_code = "0055002"
    else:
        mrg_code = "0055003"

    special_codes = {"0014010"}
    support_map = {
        "없음": None,
        "기초생활수급자": "0014003",
        "한부모가정": "0014004",
        "장애인": "0014005",
        "농업인": "0014006",
        "군인": "0014007",
        "지역인재": "0014008",
        "여성": "0014002",
        "중소기업": "0014001",
    }

    if social_support in support_map and support_map[social_support]:
        special_codes.add(support_map[social_support])

    if gender == "여성":
        special_codes.add("0014002")

    if is_sme_employee == "예":
        special_codes.add("0014001")

    user = {
        "user_id": "INPUT_USER",
        "age": age,
        "province": province,
        "region": region,
        "annual_income": float(annual_income),
        "monthly_income": float(monthly_income),
        "asset_total": float(asset_total),
        "financial_asset": float(financial_asset),
        "debt": debt_status,
        "housing_type": housing_type,
        "house_owner": house_owner,
        "spending_type": spending_type,
        "school_code": school_code,
        "job_code": job_code,
        "mrg_code": mrg_code,
        "special_codes": special_codes,
        "raw": {
            "gender": gender,
            "birth_date": str(birth_date),
            "education": education,
            "region": region,
            "monthly_income": monthly_income,
            "annual_income": annual_income,
            "house_owner": house_owner,
            "social_support": social_support,
            "spending_type": spending_type,
            "marital_status": marital_status,
            "employment_status": employment_status,
            "years_of_service": years_of_service,
            "household_type": household_type,
            "is_sme_employee": is_sme_employee,
            "asset_total": asset_total,
            "financial_asset": financial_asset,
            "debt_status": debt_status,
            "housing_type": housing_type,
            "income_type": income_type,
        }
    }
    return user
```

```python
# smart profiling 관련 함수
def categorize_profile(user: dict) -> str:
    if user["monthly_income"] <= 120 and user["house_owner"] == "무주택":
        return "지원금 탐색형"
    if user["spending_type"] == "투자형" and user["financial_asset"] >= 1500:
        return "공격적 자산형"
    if user["spending_type"] in ["절약형", "저축형"] and user["asset_total"] >= 800:
        return "안정 저축형"
    if user["housing_type"] in ["월세", "전세"] and user["house_owner"] == "무주택":
        return "주거 안정 우선형"
    return "생활 균형형"


def profile_summary(user: dict):
    return [
        f"나이 {user['age']}세 / {user['region']}",
        f"연소득 {int(user['annual_income'])}만원 / 총자산 {int(user['asset_total'])}만원",
        f"{user['housing_type']} / {user['house_owner']} / 소비성향 {user['spending_type']}",
    ]


def format_profile_text_from_input(user):
    lines = profile_summary(user)
    persona = categorize_profile(user)
    return (
        f"### 사용자 프로필 요약\n"
        f"- {lines[0]}\n"
        f"- {lines[1]}\n"
        f"- {lines[2]}\n\n"
        f"**유형화 결과:** {persona}"
    )


def make_profile_table_from_input(user):
    return pd.DataFrame([{
        "user_id": user["user_id"],
        "age": user["age"],
        "province": user["province"],
        "region": user["region"],
        "annual_income": int(user["annual_income"]),
        "monthly_income": int(user["monthly_income"]),
        "asset_total": int(user["asset_total"]),
        "financial_asset": int(user["financial_asset"]),
        "housing_type": user["housing_type"],
        "house_owner": user["house_owner"],
        "school_code": user["school_code"],
        "job_code": user["job_code"],
        "mrg_code": user["mrg_code"],
        "persona": categorize_profile(user),
    }])
```

```python
# Condition Matching 함수
def code_matches(policy_value, allowed_codes: set[str]) -> bool:
    text = safe_str(policy_value)
    if not text or text in {"nan", "None"}:
        return True
    codes = {p.strip() for p in text.split(",") if p.strip()}
    if not codes:
        return True
    if {"0013010", "0049010", "0014010"} & codes:
        return True
    return bool(codes & allowed_codes)


def marriage_matches(policy_value, mrg_code: str) -> bool:
    text = safe_str(policy_value)
    if not text or text in {"nan", "None"}:
        return True
    if "0055003" in text or "55003" in text:
        return True
    return mrg_code in text or mrg_code[-4:] in text


def age_match(policy_row: pd.Series, age: int):
    min_age = int(numeric(policy_row.get("sprtTrgtMinAge", 0), 0))
    max_age = int(numeric(policy_row.get("sprtTrgtMaxAge", 0), 0))
    if min_age == 0 and max_age == 0:
        return True
    if min_age and age < min_age:
        return False
    if max_age and age > max_age:
        return False
    return True


def income_match(policy_row: pd.Series, annual_income: float):
    income_code = safe_str(policy_row.get("earnCndSeCd", ""))
    min_amt = numeric(policy_row.get("earnMinAmt", 0), 0)
    max_amt = numeric(policy_row.get("earnMaxAmt", 0), 0)
    if "0043001" in income_code or not income_code:
        return True
    if "0043002" in income_code:
        if min_amt and annual_income < min_amt:
            return False
        if max_amt and annual_income > max_amt:
            return False
        return True
    return True


def region_match(policy_row: pd.Series, province: str):
    text = " ".join([
        safe_str(policy_row.get("sprvsnInstCdNm", "")),
        safe_str(policy_row.get("operInstCdNm", "")),
        safe_str(policy_row.get("plcyExplnCn", "")),
        safe_str(policy_row.get("plcySprtCn", "")),
        safe_str(policy_row.get("zipCd", "")),
        safe_str(policy_row.get("addAplyQlfcCndCn", "")),
    ])
    pvsn_group = safe_str(policy_row.get("pvsnInstGroupCd", ""))

    if "0054001" in pvsn_group or "54001" in pvsn_group:
        return True

    loc_keywords = {
        "서울": ["서울"],
        "경기": ["경기", "경기도"],
        "인천": ["인천"],
    }
    any_loc = any(k in text for k in [
        "서울", "경기", "경기도", "인천", "부산", "대구", "광주", "대전",
        "울산", "세종", "강원", "충북", "충남", "전북", "전남", "경북", "경남", "제주"
    ])
    if not any_loc:
        return True
    return any(k in text for k in loc_keywords.get(province, [province]))


def parse_deadline(policy_row: pd.Series):
    biz_end = safe_str(policy_row.get("bizPrdEndYmd", ""))
    if re.fullmatch(r"\d{8}", biz_end):
        try:
            return datetime.strptime(biz_end, "%Y%m%d").date()
        except Exception:
            pass

    aply = safe_str(policy_row.get("aplyYmd", ""))
    m = re.search(r"(\d{8})\s*~\s*(\d{8})", aply)
    if m:
        try:
            return datetime.strptime(m.group(2), "%Y%m%d").date()
        except Exception:
            pass

    return None


def days_to_deadline(policy_row: pd.Series):
    deadline = parse_deadline(policy_row)
    if not deadline:
        return None
    return (deadline - date.today()).days
```

```python
# 매칭 점수 계산 + 정책 추천 함수
def compute_match_score(user: dict, policy_row: pd.Series):
    score = 0
    reasons = []
    blockers = []

    if age_match(policy_row, user["age"]):
        score += 25
        reasons.append("연령 조건 부합")
    else:
        blockers.append("연령 조건 불일치")

    if region_match(policy_row, user["province"]):
        score += 20
        reasons.append("거주 지역 조건 부합 또는 전국 정책")
    else:
        blockers.append("지역 조건 불일치")

    if income_match(policy_row, user["annual_income"]):
        score += 15
        reasons.append("소득 조건 부합 또는 제한 없음")
    else:
        blockers.append("소득 조건 불일치")

    if marriage_matches(policy_row.get("mrgSttsCd"), user["mrg_code"]):
        score += 5
        reasons.append("혼인 상태 조건 부합")
    else:
        blockers.append("혼인 상태 조건 불일치")

    if code_matches(policy_row.get("schoolCd"), {user["school_code"], "0049010"}):
        score += 10
        reasons.append("학력 조건 부합")
    else:
        blockers.append("학력 조건 불일치")

    if code_matches(policy_row.get("jobCd"), {user["job_code"], "0013010"}):
        score += 10
        reasons.append("취업 조건 부합")
    else:
        blockers.append("취업 조건 불일치")

    if code_matches(policy_row.get("sbizCd"), user["special_codes"] | {"0014010"}):
        score += 10
        reasons.append("특화 조건 부합 또는 제한 없음")
    else:
        blockers.append("특화 조건 불일치")

    text = " ".join([
        safe_str(policy_row.get("plcyNm", "")),
        safe_str(policy_row.get("plcyKywdNm", "")),
        safe_str(policy_row.get("plcyExplnCn", "")),
        safe_str(policy_row.get("plcySprtCn", "")),
        safe_str(policy_row.get("lclsfNm", "")),
        safe_str(policy_row.get("mclsfNm", "")),
    ])

    if user["housing_type"] in ["월세", "전세"] and any(k in text for k in ["월세", "전세", "주거", "주택", "임차"]):
        score += 5
        reasons.append("주거 상황과 정책 목적이 유사")

    if user["spending_type"] in ["절약형", "저축형"] and any(k in text for k in ["저축", "적금", "자산", "계좌", "목돈"]):
        score += 5
        reasons.append("저축/자산형성 관련 정책")

    d = days_to_deadline(policy_row)
    if d is not None and 0 <= d <= 30:
        score += 5
        reasons.append("신청 마감 임박")

    eligible = len(blockers) == 0
    return {
        "eligible": eligible,
        "score": min(max(score, 0), 100),
        "reasons": reasons,
        "blockers": blockers,
        "deadline_days": d,
    }


def build_recommendations_for_user(user, policies_df):
    items = []

    for _, policy in policies_df.iterrows():
        status = safe_str(policy.get("plcyAprvSttsCd", ""))
        if status and status not in ["0044002", "44002", "44002.0"]:
            continue

        result = compute_match_score(user, policy)

        if result["score"] >= 45:
            items.append({
                "plcyNo": safe_str(policy.get("plcyNo", "")),
                "plcyNm": safe_str(policy.get("plcyNm", "")),
                "support_org": safe_str(policy.get("sprvsnInstCdNm") or policy.get("operInstCdNm")),
                "lclsfNm": safe_str(policy.get("lclsfNm", "")),
                "mclsfNm": safe_str(policy.get("mclsfNm", "")),
                "plcySprtCn": safe_str(policy.get("plcySprtCn", "")),
                "aplyYmd": safe_str(policy.get("aplyYmd", "")),
                "aplyUrlAddr": safe_str(policy.get("aplyUrlAddr", "")),
                "score": result["score"],
                "eligible": result["eligible"],
                "reasons": "; ".join(result["reasons"][:4]),
                "blockers": "; ".join(result["blockers"][:3]),
                "deadline_days": result["deadline_days"],
            })

    rec_df = pd.DataFrame(items)
    if not rec_df.empty:
        rec_df = rec_df.sort_values(by=["eligible", "score"], ascending=[False, False]).head(30)

    return rec_df
```

```python
# 골든타임 알림 함수
def alert_frame(rec_df: pd.DataFrame) -> pd.DataFrame:
    if rec_df.empty:
        return rec_df

    out = rec_df.copy()

    def classify(days):
        if pd.isna(days):
            return "상시/기한 정보 없음"
        if days < 0:
            return "마감"
        if days == 0:
            return "D-Day"
        if days <= 3:
            return "D-3 임박"
        if days <= 7:
            return "D-7 안내"
        if days <= 30:
            return "이번 달 체크"
        return "일반 안내"

    out["alert_type"] = out["deadline_days"].apply(classify)
    return out.sort_values(by=["deadline_days", "score"], ascending=[True, False], na_position="last")


def make_alert_table(rec_df, top_n=15):
    if rec_df.empty:
        return pd.DataFrame(columns=["plcyNm", "support_org", "score", "deadline_days", "alert_type", "aplyYmd", "aplyUrlAddr"])
    alerts = alert_frame(rec_df)
    return alerts[["plcyNm", "support_org", "score", "deadline_days", "alert_type", "aplyYmd", "aplyUrlAddr"]].head(top_n)
```

```python
# 테이블 출력용 함수
def make_recommend_table(rec_df, top_n=10):
    if rec_df.empty:
        return pd.DataFrame(columns=["plcyNm", "support_org", "score", "eligible", "reasons"])
    return rec_df[["plcyNm", "support_org", "score", "eligible", "reasons"]].head(top_n)


def make_matching_table(rec_df, top_n=10):
    if rec_df.empty:
        return pd.DataFrame(columns=[
            "plcyNm", "support_org", "lclsfNm", "mclsfNm", "score",
            "eligible", "reasons", "blockers", "deadline_days", "aplyYmd", "aplyUrlAddr"
        ])
    return rec_df[[
        "plcyNm", "support_org", "lclsfNm", "mclsfNm", "score",
        "eligible", "reasons", "blockers", "deadline_days", "aplyYmd", "aplyUrlAddr"
    ]].head(top_n)
```

```python
# 금융상품 추천
def recommend_financial_products(user, savings_df=None):
    if savings_df is None:
        return pd.DataFrame([{
            "상품명": "금융상품 데이터 수집 중",
            "추천사유": "적금/예금 데이터 연결 후 사용자 조건 기반 추천 예정"
        }])

    return pd.DataFrame()
```

```python
# 정책 문맥 생성
def build_policy_context_from_recommendations(
    rec_df: pd.DataFrame,
    max_context_tokens: int = 12000
):

    if rec_df.empty:
        return "추천된 정책이 없습니다."

    context_blocks = []
    current_tokens = 0

    for _, row in rec_df.iterrows():

        policy_text = f"""
정책명: {safe_str(row.get('plcyNm',''))}
지원내용: {safe_str(row.get('plcySprtCn',''))}
신청기간: {safe_str(row.get('aplyYmd',''))}
추천이유: {safe_str(row.get('reasons',''))}
""".strip()

        tokens = len(tokenizer.encode(policy_text))

        if current_tokens + tokens > max_context_tokens:
            break

        context_blocks.append(policy_text)
        current_tokens += tokens

    return "\n\n---\n\n".join(context_blocks)
```

```python
# gemini 프롬프트 생성
def build_personalized_prompt(question, user, rec_df):

    policy_context = build_policy_context_from_recommendations(rec_df)

    prompt = f"""
당신은 청년 정책 안내 도우미입니다.

[사용자 정보]
나이: {user['age']}
지역: {user['region']}
연소득: {user['annual_income']}
주거형태: {user['housing_type']}
소비성향: {user['spending_type']}
프로필 유형: {categorize_profile(user)}

[추천 정책]
{policy_context}

[질문]
{question}

답변은
1. 핵심 답변
2. 관련 정책
3. 신청 팁
순서로 설명하세요.
"""

    return prompt
```

```python
# gemini 답변 생성
def answer_question(question, user, rec_df, history):

    if rec_df.empty:
        return "추천 정책이 없습니다."

    client = get_client()

    prompt = build_personalized_prompt(question, user, rec_df)

    response = client.models.generate_content(
        model=GENERATION_MODEL,
        contents=prompt
    )

    answer = response.text.strip()

    history.append({"role":"user","content":question})
    history.append({"role":"assistant","content":answer})

    return answer
```

```python
# ======================================================
# 카드형 UI용 helper 함수
# ======================================================

def recommend_financial_products(user, savings_df=None):
    if savings_df is None:
        return pd.DataFrame([{
            "상품명": "금융상품 데이터 수집 중",
            "추천사유": "적금/예금 데이터 연결 후 사용자 조건 기반 추천 예정"
        }])
    return pd.DataFrame()


def make_alert_table(rec_df, selected_policies=None, top_n=15):
    if rec_df.empty:
        return pd.DataFrame(columns=["정책명", "운영기관", "매칭점수", "남은일수", "알림유형", "신청기간", "신청URL"])

    alerts = alert_frame(rec_df)

    if selected_policies:
        alerts = alerts[alerts["plcyNm"].isin(selected_policies)]

    temp = alerts[["plcyNm", "support_org", "score", "deadline_days", "alert_type", "aplyYmd", "aplyUrlAddr"]].head(top_n).copy()
    temp.columns = ["정책명", "운영기관", "매칭점수", "남은일수", "알림유형", "신청기간", "신청URL"]
    return temp


def build_policy_cards_html(rec_df: pd.DataFrame, top_n: int = 10) -> str:
    if rec_df.empty:
        return """
        <div class="empty-card">
            <div class="empty-title">추천 가능한 정책이 없습니다.</div>
            <div class="empty-desc">입력 조건을 조금 완화하거나 추가 조건을 다시 확인해보세요.</div>
        </div>
        """

    view_df = rec_df.head(top_n)
    cards = []

    for _, row in view_df.iterrows():
        policy_name = safe_str(row.get("plcyNm", ""))
        org = safe_str(row.get("support_org", ""))
        category = f"{safe_str(row.get('lclsfNm', ''))} · {safe_str(row.get('mclsfNm', ''))}".strip(" ·")
        support = safe_str(row.get("plcySprtCn", ""))
        aply_ymd = safe_str(row.get("aplyYmd", ""))
        url = safe_str(row.get("aplyUrlAddr", ""))
        score = int(row.get("score", 0)) if pd.notna(row.get("score", None)) else 0
        reasons = safe_str(row.get("reasons", ""))
        deadline = row.get("deadline_days", None)

        if pd.isna(deadline):
            deadline_badge = '<span class="badge badge-neutral">상시/기한 정보 없음</span>'
        elif deadline < 0:
            deadline_badge = '<span class="badge badge-dead">마감</span>'
        elif deadline == 0:
            deadline_badge = '<span class="badge badge-urgent">D-Day</span>'
        elif deadline <= 3:
            deadline_badge = f'<span class="badge badge-urgent">D-{int(deadline)}</span>'
        elif deadline <= 7:
            deadline_badge = f'<span class="badge badge-soon">D-{int(deadline)}</span>'
        else:
            deadline_badge = f'<span class="badge badge-normal">D-{int(deadline)}</span>'

        url_html = f'<a href="{url}" target="_blank" class="policy-link">신청 링크 바로가기</a>' if url else ""

        card_html = f"""
        <div class="policy-card">
            <div class="policy-card-header">
                <div>
                    <div class="policy-title">{policy_name}</div>
                    <div class="policy-org">{org}</div>
                </div>
                <div class="policy-score">매칭 {score}점</div>
            </div>

            <div class="policy-meta-row">
                <span class="meta-chip">{category if category else '분류 정보 없음'}</span>
                {deadline_badge}
            </div>

            <div class="policy-section">
                <div class="section-label">지원 내용</div>
                <div class="section-text">{support if support else '지원 내용 정보 없음'}</div>
            </div>

            <div class="policy-section">
                <div class="section-label">추천 이유</div>
                <div class="section-text">{reasons if reasons else '추천 이유 정보 없음'}</div>
            </div>

            <div class="policy-footer">
                <div class="policy-period"><strong>신청기간:</strong> {aply_ymd if aply_ymd else '정보 없음'}</div>
                <div>{url_html}</div>
            </div>
        </div>
        """
        cards.append(card_html)

    return "<div class='policy-grid'>" + "".join(cards) + "</div>"


def build_alert_cards_html(alerts_df: pd.DataFrame) -> str:
    if alerts_df.empty:
        return """
        <div class="empty-card">
            <div class="empty-title">선택한 정책의 알림 정보가 없습니다.</div>
            <div class="empty-desc">추천 정책을 먼저 받은 뒤, 알림 받고 싶은 정책을 선택해보세요.</div>
        </div>
        """

    cards = []

    for _, row in alerts_df.iterrows():
        title = safe_str(row.get("정책명", ""))
        org = safe_str(row.get("운영기관", ""))
        score = safe_str(row.get("매칭점수", ""))
        days = row.get("남은일수", None)
        alert_type = safe_str(row.get("알림유형", ""))
        period = safe_str(row.get("신청기간", ""))
        url = safe_str(row.get("신청URL", ""))

        if pd.isna(days):
            day_text = "기한 정보 없음"
            badge = '<span class="badge badge-neutral">상시</span>'
        elif days < 0:
            day_text = "이미 마감"
            badge = '<span class="badge badge-dead">마감</span>'
        elif days == 0:
            day_text = "오늘 마감"
            badge = '<span class="badge badge-urgent">D-Day</span>'
        else:
            day_text = f"{int(days)}일 남음"
            badge_class = "badge-urgent" if days <= 3 else ("badge-soon" if days <= 7 else "badge-normal")
            badge = f'<span class="badge {badge_class}">D-{int(days)}</span>'

        url_html = f'<a href="{url}" target="_blank" class="policy-link">신청 링크 바로가기</a>' if url else ""

        card_html = f"""
        <div class="alert-card">
            <div class="alert-top">
                <div>
                    <div class="policy-title">{title}</div>
                    <div class="policy-org">{org}</div>
                </div>
                {badge}
            </div>
            <div class="alert-info"><strong>알림 유형:</strong> {alert_type}</div>
            <div class="alert-info"><strong>남은 기간:</strong> {day_text}</div>
            <div class="alert-info"><strong>매칭 점수:</strong> {score}</div>
            <div class="alert-info"><strong>신청기간:</strong> {period if period else '정보 없음'}</div>
            <div class="alert-link-wrap">{url_html}</div>
        </div>
        """
        cards.append(card_html)

    return "<div class='alert-grid'>" + "".join(cards) + "</div>"


# ======================================================
# 실행 함수
# ======================================================

def run_profile_and_recommend(
    gender, birth_date, education, region,
    monthly_income, annual_income, house_owner, social_support,
    spending_type, marital_status, employment_status, years_of_service,
    household_type, is_sme_employee, asset_total, financial_asset,
    debt_status, housing_type, income_type
):
    user = build_user_from_form(
        gender, birth_date, education, region,
        monthly_income, annual_income, house_owner, social_support,
        spending_type, marital_status, employment_status, years_of_service,
        household_type, is_sme_employee, asset_total, financial_asset,
        debt_status, housing_type, income_type
    )

    rec_df = build_recommendations_for_user(user, policies_df)
    profile_df = make_profile_table_from_input(user)
    profile_md = format_profile_text_from_input(user)
    savings_df = recommend_financial_products(user)

    policy_names = rec_df["plcyNm"].tolist() if not rec_df.empty else []
    alert_selector_update = gr.update(
        choices=policy_names,
        value=[],
        interactive=True
    )

    empty_alert_df = pd.DataFrame(columns=["정책명", "운영기관", "매칭점수", "남은일수", "알림유형", "신청기간", "신청URL"])
    card_html = build_policy_cards_html(rec_df, top_n=10)

    return (
        profile_df,
        profile_md,
        card_html,
        savings_df,
        alert_selector_update,
        rec_df.to_dict("records"),
        format_profile_text_from_input(user),
        "추천 정책 중 알림 받고 싶은 정책을 선택해 주세요.",
        empty_alert_df,
        build_alert_cards_html(empty_alert_df)
    )


def update_selected_alerts(selected_policy_names, rec_records, profile_text):
    if not rec_records:
        empty_alert_df = pd.DataFrame(columns=["정책명", "운영기관", "매칭점수", "남은일수", "알림유형", "신청기간", "신청URL"])
        return profile_text, "먼저 정책 추천을 받아주세요.", empty_alert_df, build_alert_cards_html(empty_alert_df)

    rec_df = pd.DataFrame(rec_records)

    if not selected_policy_names:
        empty_alert_df = pd.DataFrame(columns=["정책명", "운영기관", "매칭점수", "남은일수", "알림유형", "신청기간", "신청URL"])
        return profile_text, "알림 받고 싶은 정책을 하나 이상 선택해 주세요.", empty_alert_df, build_alert_cards_html(empty_alert_df)

    alerts_df = make_alert_table(rec_df, selected_policy_names, top_n=15)
    alert_cards_html = build_alert_cards_html(alerts_df)

    if alerts_df.empty:
        summary = "선택한 정책에 대해 표시할 알림 정보가 없습니다."
    else:
        valid = alerts_df["남은일수"].dropna()
        urgent_count = ((valid >= 0) & (valid <= 7)).sum()
        if urgent_count > 0:
            summary = f"선택한 정책 중 **마감 7일 이내 정책이 {int(urgent_count)}건** 있습니다."
        else:
            summary = "선택한 정책 중 **7일 이내 마감 정책은 없습니다.**"

    return profile_text, summary, alerts_df, alert_cards_html


def chat_with_form_user(
    gender, birth_date, education, region,
    monthly_income, annual_income, house_owner, social_support,
    spending_type, marital_status, employment_status, years_of_service,
    household_type, is_sme_employee, asset_total, financial_asset,
    debt_status, housing_type, income_type,
    question, history
):
    if history is None:
        history = []

    if not question or not str(question).strip():
        return history, ""

    user = build_user_from_form(
        gender, birth_date, education, region,
        monthly_income, annual_income, house_owner, social_support,
        spending_type, marital_status, employment_status, years_of_service,
        household_type, is_sme_employee, asset_total, financial_asset,
        debt_status, housing_type, income_type
    )

    rec_df = build_recommendations_for_user(user, policies_df)

    conversation_history = []
    for item in history:
        if isinstance(item, dict):
            conversation_history.append(item)

    try:
        answer = answer_question(question, user, rec_df, conversation_history)
    except Exception as e:
        answer = f"질의응답 중 오류 발생: {e}"

    history.append({"role": "user", "content": question})
    history.append({"role": "assistant", "content": answer})

    return history, ""


# ======================================================
# 카드형 Gradio UI
# ======================================================

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="blue",
        secondary_hue="sky",
        neutral_hue="slate"
    ),
    css="""
    .gradio-container {
        max-width: 1280px !important;
        margin: 0 auto !important;
        padding-top: 12px !important;
    }

    .hero-box {
        background: linear-gradient(135deg, #eef6ff 0%, #f8fbff 100%);
        border: 1px solid #dbeafe;
        border-radius: 20px;
        padding: 28px 24px;
        margin-bottom: 18px;
    }

    .section-card {
        background: white;
        border: 1px solid #e5e7eb;
        border-radius: 18px;
        padding: 18px;
        box-shadow: 0 2px 10px rgba(15, 23, 42, 0.04);
    }

    .section-title {
        font-size: 20px;
        font-weight: 700;
        margin-bottom: 6px;
        color: #0f172a;
    }

    .section-desc {
        font-size: 14px;
        color: #475569;
        margin-bottom: 14px;
    }

    .sub-card {
        background: #fafafa;
        border: 1px solid #e5e7eb;
        border-radius: 16px;
        padding: 14px 14px 8px 14px;
    }

    .result-title {
        font-size: 18px;
        font-weight: 700;
        margin-bottom: 6px;
        color: #111827;
    }

    .result-desc {
        font-size: 13px;
        color: #6b7280;
        margin-bottom: 10px;
    }

    .policy-grid {
        display: grid;
        grid-template-columns: repeat(auto-fit, minmax(320px, 1fr));
        gap: 14px;
        margin-top: 8px;
    }

    .policy-card {
        background: #ffffff;
        border: 1px solid #dbe4f0;
        border-radius: 18px;
        padding: 18px;
        box-shadow: 0 4px 14px rgba(15, 23, 42, 0.05);
    }

    .policy-card-header {
        display: flex;
        justify-content: space-between;
        align-items: flex-start;
        gap: 12px;
        margin-bottom: 12px;
    }

    .policy-title {
        font-size: 18px;
        font-weight: 700;
        color: #0f172a;
        line-height: 1.35;
    }

    .policy-org {
        font-size: 13px;
        color: #64748b;
        margin-top: 4px;
    }

    .policy-score {
        background: #eff6ff;
        color: #1d4ed8;
        border-radius: 12px;
        padding: 6px 10px;
        font-size: 13px;
        font-weight: 700;
        white-space: nowrap;
    }

    .policy-meta-row {
        display: flex;
        flex-wrap: wrap;
        gap: 8px;
        margin-bottom: 12px;
    }

    .meta-chip {
        background: #f1f5f9;
        color: #334155;
        padding: 6px 10px;
        border-radius: 999px;
        font-size: 12px;
        font-weight: 600;
    }

    .badge {
        padding: 6px 10px;
        border-radius: 999px;
        font-size: 12px;
        font-weight: 700;
    }

    .badge-urgent {
        background: #fee2e2;
        color: #b91c1c;
    }

    .badge-soon {
        background: #fef3c7;
        color: #b45309;
    }

    .badge-normal {
        background: #dcfce7;
        color: #15803d;
    }

    .badge-neutral {
        background: #e2e8f0;
        color: #475569;
    }

    .badge-dead {
        background: #e5e7eb;
        color: #6b7280;
    }

    .policy-section {
        margin-bottom: 12px;
    }

    .section-label {
        font-size: 12px;
        font-weight: 700;
        color: #475569;
        margin-bottom: 4px;
        text-transform: uppercase;
        letter-spacing: 0.02em;
    }

    .section-text {
        font-size: 14px;
        color: #1f2937;
        line-height: 1.5;
    }

    .policy-footer {
        display: flex;
        flex-direction: column;
        gap: 8px;
        margin-top: 12px;
        padding-top: 12px;
        border-top: 1px solid #e5e7eb;
    }

    .policy-period {
        font-size: 13px;
        color: #374151;
    }

    .policy-link {
        display: inline-block;
        color: #2563eb;
        text-decoration: none;
        font-weight: 700;
        font-size: 13px;
    }

    .policy-link:hover {
        text-decoration: underline;
    }

    .alert-grid {
        display: grid;
        grid-template-columns: repeat(auto-fit, minmax(320px, 1fr));
        gap: 14px;
        margin-top: 8px;
    }

    .alert-card {
        background: #ffffff;
        border: 1px solid #dbe4f0;
        border-radius: 18px;
        padding: 18px;
        box-shadow: 0 4px 14px rgba(15, 23, 42, 0.05);
    }

    .alert-top {
        display: flex;
        justify-content: space-between;
        align-items: flex-start;
        gap: 10px;
        margin-bottom: 12px;
    }

    .alert-info {
        font-size: 14px;
        color: #334155;
        margin-bottom: 6px;
    }

    .alert-link-wrap {
        margin-top: 10px;
    }

    .empty-card {
        border: 1px dashed #cbd5e1;
        background: #f8fafc;
        border-radius: 18px;
        padding: 24px;
        text-align: center;
    }

    .empty-title {
        font-size: 17px;
        font-weight: 700;
        color: #334155;
        margin-bottom: 6px;
    }

    .empty-desc {
        font-size: 14px;
        color: #64748b;
    }

    .footer-note {
        color: #64748b;
        font-size: 13px;
    }
    """
) as demo:

    rec_state = gr.State([])
    profile_text_state = gr.State("")

    with gr.Column(elem_classes=["hero-box"]):
        gr.Markdown("""
        # 청년 맞춤 금융·정책 추천 서비스
        사용자가 직접 조건을 입력하면 **지금 신청 가능한 청년 정책**을 추천하고,  
        **알림 받고 싶은 정책만 선택해서 D-Day 관리**할 수 있도록 구성한 서비스입니다.
        """)

    with gr.Row(equal_height=False):
        with gr.Column(scale=5):
            with gr.Column(elem_classes=["section-card"]):
                gr.Markdown('<div class="section-title">1. 사용자 정보 입력</div>')
                gr.Markdown('<div class="section-desc">기본 정보는 필수 입력이며, 추가 조건은 선택 입력입니다. 추가 조건을 넣을수록 추천 정확도가 높아집니다.</div>')

                with gr.Column(elem_classes=["sub-card"]):
                    gr.Markdown("### 기본 정보")

                    with gr.Row():
                        gender = gr.Radio(["남성", "여성"], label="성별", value="여성")
                        birth_date = gr.Textbox(label="생년월일", value="2000-01-01", placeholder="YYYY-MM-DD")
                        education = gr.Dropdown(
                            ["고교 졸업", "대학 재학", "대졸 예정", "대학 졸업", "석박사", "기타"],
                            label="학력", value="대학 졸업"
                        )

                    region = gr.Textbox(label="지역(시/군/구)", value="서울관악구", placeholder="예: 서울관악구")

                    with gr.Row():
                        monthly_income = gr.Number(label="월 소득(만원)", value=200)
                        annual_income = gr.Number(label="연 소득(만원)", value=2400)

                    with gr.Row():
                        house_owner = gr.Radio(["무주택", "주택보유"], label="주택 여부", value="무주택")
                        social_support = gr.Dropdown(
                            ["없음", "기초생활수급자", "한부모가정", "장애인", "농업인", "군인", "지역인재", "여성", "중소기업"],
                            label="사회배려자 대상 여부", value="없음"
                        )

                with gr.Accordion("추가 조건 입력 (선택)", open=False):
                    with gr.Column(elem_classes=["sub-card"]):
                        gr.Markdown("### 추가 조건")

                        with gr.Row():
                            spending_type = gr.Dropdown(["절약형", "소비형", "저축형", "투자형"], label="소비 성향", value="저축형")
                            marital_status = gr.Radio(["미혼", "기혼"], label="결혼 여부", value="미혼")
                            employment_status = gr.Dropdown(
                                ["취업", "취업준비중", "학생", "프리랜서", "자영업", "일용근로", "예비창업", "단기근로", "영농종사", "기타"],
                                label="고용 상태", value="취업준비중"
                            )

                        with gr.Row():
                            years_of_service = gr.Dropdown(["0년", "1년 미만", "1~3년", "3~5년", "5년 이상"], label="근속 기간", value="0년")
                            household_type = gr.Dropdown(["1인 가구", "부모 동거", "부부", "다자녀"], label="가구 형태", value="1인 가구")
                            is_sme_employee = gr.Radio(["예", "아니오"], label="중소기업 재직 여부", value="아니오")

                        with gr.Row():
                            asset_total = gr.Number(label="총 자산 규모(만원)", value=500)
                            financial_asset = gr.Number(label="금융자산 규모(만원)", value=200)
                            debt_status = gr.Dropdown(["없음", "학자금대출", "신용대출", "기타"], label="부채 여부", value="없음")

                        with gr.Row():
                            housing_type = gr.Dropdown(["월세", "전세", "자가", "부모동거", "기숙사"], label="주거 형태", value="월세")
                            income_type = gr.Dropdown(["근로소득", "사업소득", "무소득", "기타"], label="소득 유형", value="무소득")

                with gr.Row():
                    submit_btn = gr.Button("입력 완료 · 정책 추천 받기", variant="primary", size="lg")
                    reset_btn = gr.Button("입력 초기화", variant="secondary")

        with gr.Column(scale=4):
            with gr.Column(elem_classes=["section-card"]):
                gr.Markdown('<div class="section-title">2. 사용자 요약</div>')
                gr.Markdown('<div class="section-desc">입력 완료 버튼을 누르면 사용자 조건을 분석해 금융 MBTI와 추천 방향을 보여줍니다.</div>')

                profile_md = gr.Markdown("아직 사용자 분석 전입니다.")
                profile_df = gr.Dataframe(label="프로파일링 결과", interactive=False)

            with gr.Column(elem_classes=["section-card"]):
                gr.Markdown('<div class="section-title">3. 추천 금융상품(예정)</div>')
                gr.Markdown('<div class="section-desc">적금/예금 데이터 연결 후 사용자 조건에 맞는 금융상품까지 함께 추천할 수 있습니다.</div>')

                savings_df = gr.Dataframe(label="추천 금융상품(적금)", interactive=False)

    with gr.Tabs():
        with gr.Tab("추천 정책"):
            with gr.Column(elem_classes=["section-card"]):
                gr.Markdown('<div class="result-title">사용자 조건 필터링 결과</div>')
                gr.Markdown('<div class="result-desc">입력된 조건을 바탕으로 현재 신청 가능성이 높은 정책을 카드형으로 보여줍니다.</div>')

                preview_cards = gr.HTML(label="추천 정책 카드")

        with gr.Tab("골든타임 알림"):
            with gr.Column(elem_classes=["section-card"]):
                gr.Markdown('<div class="result-title">신청 골든타임 알림</div>')
                gr.Markdown('<div class="result-desc">추천된 정책 중 알림 받고 싶은 정책만 골라서 D-Day를 확인할 수 있습니다.</div>')

                alert_profile_md = gr.Markdown("먼저 정책 추천을 받아주세요.")

                alert_selector = gr.Dropdown(
                    choices=[],
                    value=[],
                    multiselect=True,
                    label="알림 받고 싶은 정책 선택"
                )

                alert_summary_md = gr.Markdown("아직 선택된 정책이 없습니다.")
                alerts_df = gr.Dataframe(label="선택 정책 D-Day 알림 표", interactive=False)
                alert_cards = gr.HTML(label="선택 정책 알림 카드")

        with gr.Tab("AI 정책 상담"):
            with gr.Column(elem_classes=["section-card"]):
                gr.Markdown('<div class="result-title">AI 질의형 대화</div>')
                gr.Markdown('<div class="result-desc">추천된 정책을 기반으로 지금 신청 가능한 정책, 주거 지원, 자산 형성 정책 등을 자연어로 물어볼 수 있습니다.</div>')

                chatbot = gr.Chatbot(type="messages", label="정책 상담 챗봇", height=420)

                with gr.Row():
                    chat_input = gr.Textbox(
                        label="질문 입력",
                        placeholder="예: 나 이번 달에 신청할 수 있는 지원금 있어?",
                        scale=8
                    )
                    send_btn = gr.Button("질문하기", variant="primary", scale=1)

                clear_btn = gr.Button("대화 초기화", variant="secondary")

    gr.Markdown('<div class="footer-note">※ 현재 금융상품 추천은 데이터 연결 전 단계이며, 청년 정책 추천/알림/상담 기능 중심으로 동작합니다.</div>')

    common_inputs = [
        gender, birth_date, education, region,
        monthly_income, annual_income, house_owner, social_support,
        spending_type, marital_status, employment_status, years_of_service,
        household_type, is_sme_employee, asset_total, financial_asset,
        debt_status, housing_type, income_type
    ]

    submit_btn.click(
        fn=run_profile_and_recommend,
        inputs=common_inputs,
        outputs=[
            profile_df,
            profile_md,
            preview_cards,
            savings_df,
            alert_selector,
            rec_state,
            alert_profile_md,
            alert_summary_md,
            alerts_df,
            alert_cards
        ]
    ).then(
        fn=lambda text: text,
        inputs=profile_md,
        outputs=profile_text_state
    )

    alert_selector.change(
        fn=update_selected_alerts,
        inputs=[alert_selector, rec_state, profile_text_state],
        outputs=[alert_profile_md, alert_summary_md, alerts_df, alert_cards]
    )

    send_btn.click(
        fn=chat_with_form_user,
        inputs=common_inputs + [chat_input, chatbot],
        outputs=[chatbot, chat_input]
    )

    chat_input.submit(
        fn=chat_with_form_user,
        inputs=common_inputs + [chat_input, chatbot],
        outputs=[chatbot, chat_input]
    )

    clear_btn.click(
        fn=lambda: [],
        inputs=None,
        outputs=chatbot
    )

    reset_btn.click(
        fn=lambda: (
            "여성", "2000-01-01", "대학 졸업", "서울관악구",
            200, 2400, "무주택", "없음",
            "저축형", "미혼", "취업준비중", "0년",
            "1인 가구", "아니오", 500, 200,
            "없음", "월세", "무소득",
            "아직 사용자 분석 전입니다.", pd.DataFrame(), pd.DataFrame(),
            gr.update(choices=[], value=[]),
            [], "먼저 정책 추천을 받아주세요.", "아직 선택된 정책이 없습니다.", pd.DataFrame(),
            build_alert_cards_html(pd.DataFrame()), []
        ),
        inputs=None,
        outputs=[
            gender, birth_date, education, region,
            monthly_income, annual_income, house_owner, social_support,
            spending_type, marital_status, employment_status, years_of_service,
            household_type, is_sme_employee, asset_total, financial_asset,
            debt_status, housing_type, income_type,
            profile_md, profile_df, savings_df,
            alert_selector, rec_state, alert_profile_md, alert_summary_md, alerts_df,
            alert_cards, chatbot
        ]
    )

demo.launch(inline=False, share=False)
```
