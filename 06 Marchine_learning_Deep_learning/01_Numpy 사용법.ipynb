# Numpy를 사용해야 하는 이유
* 행렬 연산이 쉽고 빠르기 때문

```python
#  스칼라 연산
a = 10
b = 5
scalar = a + b
scalar
```

```python
# 1차원 벡터의 연산
v1 = [1,2,3,4,5]
# v1 요소에 10을 곱한 리스트로 만드세요
mul10 = []
for num in v1:
    mul10.append(num*10)
mul10
```

```python
# 3차원 데이터에 10을 곱하는 연산
result3 = []
for item1 in d3:
    temp2 = []
    for item2 in item1:
        temp1 = []
        for item3 in item2:
            temp1.append(item3 * 10)
        temp2.append(temp1)
    result3.append(item2)
result3 = np.array(result3)
print(result3, "\n", result3.ndim, result3.shape)
```

# numpy에서 자주 사용하는 메서드
* .ndim: 몇 차원인지 확인
* .shape: 몇 차원, 몇 행, 몇 열인지 구조를 확인
* .size: 배열 안에 있는 요소의 총 개수가 몇 개인지
* .dtype: 데이터 타입이 무엇인지

# ndarray 만들기
* numpy의 기본 자료형 ndarray
* np.array(리스트/튜플)
* 자료형은 반드시 1가지만 넣음

```python
e1 = np.array([1,2,3,4,5])
e1
```

```python
e2 = np.array((1,2,3,4,5))
e2

# np.array()는 리스트([])든 튜플(())이든 순서 있는 자료형”이면 그대로 배열로 바꿔줌
```

```python
e3 = np.array((1,2,3,4,5))
print(e3, e3.dtype)
```

```python
e4 = np.array((1, 2, 3.0, "4", 5))
print(e4, e4.dtype)
```

# .reshape(행, 열)
* 벡터를 원하는 크기의 다차원 배열로 변환

```python
a4 = a3.reshape(5, 2, 2, 5)
print(a4.ndim, a4.shape)
```
