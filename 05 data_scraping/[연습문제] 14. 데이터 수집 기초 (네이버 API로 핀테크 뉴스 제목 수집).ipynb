### 1) 뉴스 데이터 수집

```python
def search_news(query, display=100, start=1):
    url = "https://openapi.naver.com/v1/search/news.json"
    headers = {"X-Naver-Client-Id": client_id,
        "X-Naver-Client-Secret": client_secret}
    params = {"query": query,
                "display": display,
                "start": start,
                "sort": "date"}
    response = requests.get(url, headers=headers, params=params)
    return response.json()

data = search_news("핀테크")
```

### 2) 페이지네이션: 여러 페이지에서 가져온 데이터를 하나로 모음

```python
all_news = []

for start in range(1, 1000, 100):
    data = search_news("핀테크", start=start)

    for item in data["items"]:
        all_news.append(item)
        
print(len(all_news))
```

```python
df = pd.DataFrame(all_news)
df
```

### 3. 데이터 정제

```python
import re
```

```python
# 1. 컬럼 삭제
df = df.drop(columns=["originallink"])

# 2. HTML 태그 제거 함수 정의
def remove_html(text):
    return re.sub('<.*?>', '', text)

# 3. 함수 적용
df["title"] = df["title"].apply(remove_html)
df["description"] = df["description"].apply(remove_html)

# 4. 확인
df
```
